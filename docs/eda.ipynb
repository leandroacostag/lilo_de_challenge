# EDA for Lilo DE Challenge

```python
# Loading data
import json

products = json.load(open("../data/products.json"))
products[0]
```

### Unit of measures

```python
unit_of_measures = {product["unit_of_measure"] for product in products}

print(f"unit_of_measures: {unit_of_measures}")
print(f"len(unit_of_measures): {len(unit_of_measures)}")
```

### Region availability

```python
region_availability = set()

for product in products:
    region_availability.update(product["region_availability"])

print(f"region_availability: {region_availability}")
print(f"len(region_availability): {len(region_availability)}")
```

### Supplier rating

```python
supplier_ratings = [product["supplier_rating"] for product in products]
supplier_ratings_dtypes = {type(rating) for rating in supplier_ratings}
print(f"supplier_ratings: {supplier_ratings}")
print(f"supplier_ratings_dtypes: {supplier_ratings_dtypes}")
print(f"len(supplier_ratings): {len(supplier_ratings)}")
print(f"min(supplier_ratings): {min(supplier_ratings)}")
print(f"max(supplier_ratings): {max(supplier_ratings)}")
```

### Inventory Status

```python
inventory_statuses = {product["inventory_status"] for product in products}

print(f"inventory_statuses: {inventory_statuses}")
print(f"len(inventory_statuses): {len(inventory_statuses)}")
```

### Bulk Pack Size

```python
bulk_pack_sizes = {product["bulk_pack_size"] for product in products}

print(f"bulk_pack_sizes: {bulk_pack_sizes}")
print(f"len(bulk_pack_sizes): {len(bulk_pack_sizes)}")
```

### Category

```python
categories = {product["category"] for product in products}

print(f"categories: {categories}")
print(f"len(categories): {len(categories)}")
```

### Vendor

```python
vendors = {product["vendor"] for product in products}

print(f"vendors: {vendors}")
print(f"unique vendors: {len(vendors)}")
print(f"null vendors: {sum(1 for product in products if product['vendor'] is None)}")
```

### Attributes

```python
attribute_fields = set()
for product in products:
    attribute_fields.update(product["attributes"].keys())

print(f"attribute_fields: {attribute_fields}")
print(f"len(attribute_fields): {len(attribute_fields)}")
```

#### Exploring diameter

```python
[att for att in attribute_fields if any(s in att.lower() for s in ["dia", "mm", "diam", "diameter", "iad"])]
```
